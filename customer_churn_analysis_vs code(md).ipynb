# Customer Churn Analysis
This notebook performs an initial exploration of a customer churn dataset: loading the data, inspecting its structure, and checking data quality (duplicates and missing values).

## 1. Import Libraries
Import `pandas`, the library used for data loading and manipulation.

```python
import pandas as pd
```

## 2. Load the Dataset
Read the `customer churn.csv` file into a pandas DataFrame called `df`.

```python
df=pd.read_csv("customer churn.csv")
```

## 3. Dataset Overview
### 3.1 Structure and Data Types
`df.info()` shows the number of rows, column names, data types, and non-null counts for each column.

```python
df.info()
```

### 3.2 First Five Rows
`df.head()` displays the first 5 records to give a quick look at the data.

```python
df.head()
```

### 3.3 Last Five Rows
`df.tail()` displays the last 5 records to confirm the data ends as expected.

```python
df.tail()
```

## 4. Save a Copy of the Data
Export the DataFrame to a new file named `car.csv`. `index=False` prevents the row index from being written as an extra column.

```python
df.to_csv("car.csv", index=False)
```

## 5. Statistical Summary
`df.describe()` gives summary statistics (count, mean, std, min, quartiles, max) for the numerical columns.

```python
df.describe()
```

### 5.1 Shape of the Data
`df.shape` returns the dataset size as (number of rows, number of columns).

```python
df.shape
```

## 6. Data Quality Checks
### 6.1 Duplicate Records
Count the number of fully duplicated rows. A result of 0 means there are no duplicates.

```python
df.duplicated().sum()
```

### 6.2 Missing Values (Boolean View)
`df.isnull()` returns a True/False table showing where values are missing.

```python
df.isnull()
```

### 6.3 Missing Values (Count per Column)
Sum the missing values in each column to see which columns need cleaning.

```python
df.isnull().sum()
```
